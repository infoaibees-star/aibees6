### How many people having word chief in their job title?

```python
def chief_string(title):
    if 'chief' in title.lower(): #CHIEF, Chief, chief, cHiEf, chIEf, etc.
        return True
    else:
        return False
```

```python
sal['JobTitle'].apply(lambda x : chief_string(x))
```

```python
sum(sal['JobTitle'].apply(lambda x : chief_string(x)))
```
