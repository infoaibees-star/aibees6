```python
score = 80
if score >= 90:                 # >, <, >=, <=, ==, !=
    print("Grade: A")
```

```python
score = 80
if score >= 90:
    print("Grade: A")
else:
    print("Grade: B")    
```

# Exception handling
- try and except block is used to handle exceptions in python. It allows you to catch and handle errors gracefully, preventing your program from crashing.

```python
def even_or_odd(num):
    """
    Determines if a number is even or odd.
    It requires a numeric input and returns a string indicating whether the number is even or odd.
    """
    try:
        if num % 2 == 0:
            return f"{num} is even"
        else:
            return f"{num} is odd"
    except TypeError:
        return "Please enter a valid number."
```

```python
even_or_odd(6)
```

```python
even_or_odd("6")
```

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")    
```

# Class and OOP
- class LLM : Blueprint or Building plan
- __init__: constructor or tools/materials, construction crew
- summarize: Action the building can perform
- gemini = LLM("Gemini) :Actual building object
- gemini.summarize : Method calling or using the building
- self : refer to instance itself or which building we are referring to

```python
class LLM:
    def __init__(self, model_name):
        self.model_name = model_name
    
    def summarize(self, text):
        return f"Summarizing text using {self.model_name}: {text[:50]}..."  # Just a placeholder implementation


gemini = LLM("Gemini")
print(gemini.summarize("This is a sample text for summarization."))

```

```python
class BankAccount:
    """
    Blue print for creating individual bank accounts
    Each account has a name and its own unique balance 
    """
    # Constructor
    def __init__(self, owner_name, initial_balance=0):
        self.name = owner_name
        self.balance = initial_balance
        print(f"Account created for {self.name} with initial balance of {self.balance}")

    # Method to deposit money
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount}. New balance is {self.balance}")
        else:
            print("Deposit amount must be positive.")

    # Method to withdraw money
    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance is {self.balance}")
        else:
            print("Withdrawal amount must be positive and less than or equal to the current balance.")   


account_alice = BankAccount("Alice", 1000)
account_alice.deposit(500)
account_alice.withdraw(300)
print(f"Alice's final balance: {account_alice.balance}")
```
